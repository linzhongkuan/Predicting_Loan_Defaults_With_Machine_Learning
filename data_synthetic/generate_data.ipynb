```python
# Import required modules
import numpy as np
import pandas as pd
from faker import Faker
```

```python
def generate_synthetic_dataset(n_rows):
    # Initialize Faker with appropriate locale
    fake = Faker(locale='en')

    applicant_id = [] # String, Primary key
    income = [] # Integer, HKD
    loan_amount = [] # Integer, HKD
    term = [] # Integer, Repayment period
    credit_history = [] # Boolean, Good/Bad
    default = [] # Boolean, Yes/No

    # Term durations range from 6-60 months, in multiples of 6.
    term_durations = np.arange(6,61,6)

    # Generate dataset with a loop
    for _ in range(n_rows):
        # Generate unique string identifier with faker for applicant id
        applicant_id.append(fake.uuid4())

        # Generate values for income assuming normal distribution
            # Using mean income in HK for Q1 2026, $21_000
            # Source: HK Government
                # https://www.censtatd.gov.hk/en/wbr.html?ecode=B10500012026QQ01&scode=200
        # Limitation: 
            # Income distributions are usually right-skewed, so this generation does not accurately 
            # reflect a real-world scenario.
            # Because the dataset is fully synthetic and used for analysis of a hypothetical scenario,
            # this can be overlooked.
        generated_income = int(np.random.normal(loc=21_000, scale=7_000))
        # Some banks require a minimum wage of 5000 HKD to apply for personal loans
        income.append(max(5000, generated_income))

        # Generate random loan amount
        generated_principal = np.random.randint(100_000, 1_000_000)
        loan_amount.append(int(generated_principal))

        # Generate term duration
        generated_term = np.random.choice(term_durations)
        term.append(generated_term)

        # Generate credit history (1=clean, no past defaults; 0 = past defaults or many missed payments)
        credit_score = np.random.choice([1,0])
        credit_history.append(credit_score)

        # Generate default (1=yes, 0=no)
        if credit_score == 0 or generated_income > int(generated_principal/generated_term):
            default.append(np.random.choice([1,0], p=[0.8, 0.2]))
        else:
            default.append(np.random.choice([1,0], p=[0.1, 0.9]))

    # Store dataset in pandas datafram
    data = pd.DataFrame(
        {
            'applicant_id': applicant_id,
            'income': income,
            'loan_amount': loan_amount,
            'term': term,
            'credit_history': credit_history,
            'default': default
        }
    )

    return data


```

```python
# Create synthetic dataset
synthetic_df = generate_synthetic_dataset(n_rows=2000)
synthetic_df
```

```python
# Export to csv
synthetic_df.to_csv("dataset.csv",index=False)
```
